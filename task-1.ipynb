## Create a DataFrame

#Create a DataFrame with columns: Name, Age, and Marks
#Include 4 student records with at least one missing value.
#(Hint: Use a dictionary and pd.DataFrame())

```python
import pandas as pd
import numpy as np
data = {
    "Name":  ["sahal", "Ravi", "arjun", "Kiran"],
    "Age":   [20, 21, np.nan, 22],
    "Marks": [85, 78, 92, np.nan]
}

df = pd.DataFrame(data)

print(df)
```

##  Add a new column

#Add a new column "Grade" with letter grades (A, B, C, etc.)

```python

def get_grade(marks):
    if pd.isnull(marks):
        return "N/A"   
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Marks"].apply(get_grade)

print(df)
```

## Access specific columns

#Display only the 'Name' and 'Marks' columns.

```python
print(df[["Name", "Marks"]])
```

## Count missing values

#Count the number of missing values in each column.
#(Hint: df.isnull().sum())

```python
print(df.isnull().sum())
```

## Handle missing values

#Replace all missing values in 'Age' with the mean age.
#(Hint: df['Age'].fillna(df['Age'].mean(), inplace=True))

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

## Modify a specific cell

#Change the Name in the 3rd row (index 2) to 'Sonu'.
#(Hint: df.loc[2, 'Name'] = 'Sonu')

```python
df.loc[2, "Name"] = "Sonu"

print(df)
```

## Add new rows to the DataFrame

#Add two new rows:
#{'Name': 'Kumar', 'Age': 20, 'Marks': 90}
#{'Name': 'Rahul', 'Age': 26, 'Marks': 85}
#(Hint: Use pd.concat)

```python
new_rows = pd.DataFrame([
    {"Name": "Kumar", "Age": 20, "Marks": 90},
    {"Name": "Rahul", "Age": 26, "Marks": 85}
])

df = pd.concat([df, new_rows], ignore_index=True)

print(df)
```

## Aggregate functions

#Calculate and display:
#Mean Age
#Minimum Age
#Maximum Age
#Count of Age values






```python
print("Mean Age:", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
print("Count of Age values:", df["Age"].count())
```
