### productA = [120, 150, 170, 190, 210, 230]

# 2️⃣ BAR GRAPH: Total Sales by Country

### Task: Create a bar chart showing total sales for each country.
### Hint:
###   1. Add a 'Sales' column: df['Sales'] = df['Quantity'] * df['Price']
###   2. Group by 'Country' and sum 'Sales'
###  3. Sort values in descending order
###  4. Use plt.bar() to plot
###  5. Display the sales value above each bar


```python
import pandas as pd
import matplotlib.pyplot as plt

dataset = {
    'Country': ['India', 'USA', 'UK', 'India', 'USA', 'UK'],
    'Quantity': [10, 20, 15, 5, 10, 8],
    'Price': [100, 200, 150, 120, 180, 100]
}

df = pd.DataFrame(dataset)

df['Sales'] = df['Quantity'] * df['Price']

country_sales = (
    df.groupby('Country')['Sales']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
bars = plt.bar(country_sales.index, country_sales.values, color='steelblue')

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,  
        height,                             
        f'{height:,.0f}',                   
        ha='center', va='bottom'
    )

plt.title('Total Sales by Country')
plt.xlabel('Country')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# 3️⃣ LINE GRAPH: Daily Sales Trend

### Task: Plot a line graph showing total sales per day.
### Hint:
###   1. Convert 'OrderDate' to datetime: df['OrderDate'] = pd.to_datetime(df['OrderDate'])
###   2. Group by 'OrderDate' and sum 'Sales'
###   3. Use plt.plot() with markers and line style
###   4. Label X-axis, Y-axis, and add a title


```python
import pandas as pd
import matplotlib.pyplot as plt

dataset = {
    'OrderDate': [
        '2026-10-20', '2026-10-20',
        '2026-10-21', '2026-10-21',
        '2026-10-22', '2026-10-23'
    ],
    'Sales': [1000, 1500, 2000, 1200, 1800, 2500]
}

df = pd.DataFrame(dataset)

df['OrderDate'] = pd.to_datetime(df['OrderDate'])

daily_sales = df.groupby('OrderDate')['Sales'].sum().sort_index()

plt.figure(figsize=(12, 6))
plt.plot(
    daily_sales.index,
    daily_sales.values,
    marker='o',
    linestyle='-',
    color='steelblue',
    linewidth=2,
    markersize=5
)

plt.title('Daily Sales Trend')
plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# 4️⃣ LINE GRAPH: Category-wise Average Price

### Hint:
###   1. Group by 'Category' and calculate mean of 'Price'
###   2. Use plt.plot() with markers, colors, and line style
###  3. Label axes and add grid lines


```python
import pandas as pd
import matplotlib.pyplot as plt

dataset = {
    'Category': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],
    'Price': [50000, 20000, 15000, 60000, 25000, 18000]
}

df = pd.DataFrame(dataset)

avg_price = df.groupby('Category')['Price'].mean()

plt.figure(figsize=(10, 6))
plt.plot(
    avg_price.index,
    avg_price.values,
    marker='s',
    linestyle='--',
    color='darkorange',
    linewidth=2,
    markersize=8
)

plt.title('Average Product Price by Category')
plt.xlabel('Category')
plt.ylabel('Average Price')
plt.grid(True, linestyle=':', alpha=0.6)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 2️⃣ Plot a scatter graph comparing 'Age' and 'Annual Income' from the dataset.
#     - Use different colors or markers for different 'Gender' categories.



```python
import pandas as pd
import matplotlib.pyplot as plt

df_income = pd.DataFrame({
    'Age': [22, 25, 27, 29, 30, 32, 35, 37, 40, 45],
    'Annual Income': [25000, 27000, 30000, 33000, 36000, 40000, 45000, 48000, 52000, 60000],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female']
})

styles = {
    'Male':   {'color': 'steelblue',  'marker': 'o'},
    'Female': {'color': 'darkorange', 'marker': 's'}
}

plt.figure(figsize=(8, 6))

for gender, style in styles.items():
    subset = df_income[df_income['Gender'] == gender]
    plt.scatter(
        subset['Age'],
        subset['Annual Income'],
        color=style['color'],
        marker=style['marker'],
        s=90,
        edgecolors='black',
        label=gender
    )

plt.xlabel('Age')
plt.ylabel('Annual Income')
plt.title('Age vs Annual Income by Gender')
plt.legend(title='Gender')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
