## matplotlib

# data visualization using python 2d plotting library

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(matplotlib.__version__)
```

```python
x=np.array([2022,2023,2024,2025])
y=np.array([0,0,14,25])

plt.plot(x,y, marker='o',
         markersize=8,
         markerfacecolor='red',
         markeredgecolor='red',
         linestyle='solid',
         linewidth=2,
         color='green')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([2022,2023,2024,2025])
y1=np.array([0,0,14,25])
y2=np.array([0,5,10,20])
y3=np.array([0,0,11,22])
y4=np.array([0,0,110,23])

line_style = dict(marker='o',
           markersize=10,
					markerfacecolor='red',
					markeredgecolor='red',
					linestyle='solid',
					linewidth=2,
					color='blue')

plt.plot(x, y1, **line_style)

plt.plot(x, y2, **line_style)

plt.plot(x, y3, **line_style)

plt.plot(x, y4, **line_style)

plt.show()
          

```

## grid

#  grid() = grid() function in Matplotlib is used to enhance the readability and interpretability 
 of plots by adding a grid of horizontal and/or vertical lines. These lines serve as visual aids, 
 making it easier to analyze and understand the data presented in a graph.
'''

```python
x=np.array([1, 2, 3, 4, 5])
y=np.array([5, 10, 15, 20, 25])

plt.grid(linewidth=1,
         color='red',
         linestyle=':')
plt.plot(x, y, marker='v')
plt.show()
```

# bar chart

```python
fruits = np.array(['apples', 'bananas', 'cherries', 'dates'])
sales = np.array([500, 350, 300, 450])


line_style = dict(fontsize=15,
                  family='arial',
                  fontweight='bold',
                  color='blue')

plt.bar(fruits, sales, color='skyblue')
plt.title('fruits sales', fontsize=15,color='skyblue')
plt.xlabel('fruits', **line_style)
plt.ylabel('sales', **line_style)
plt.show()
```

## pie chart

```python
x=np.array([25,50,20,5,8])
mylabels = ["apples", "bananas", "cherries", "dates", "oranges"]

plt.pie(x, labels= mylabels,
        autopct="%1.2f%%"
)

plt.title('fruit sales')
plt.show()
```

## scatter plot

```python
x1 = np.array([0, 1, 5, 6, 7, 4, 4, 3, 8,]) # student study  hour
y1 = np.array([10, 20, 50, 65, 75, 60, 60, 55, 85]) # stu

plt.title('test scores')
plt.xlabel('hours studied')
plt.ylabel('marks')

plt.scatter(x1, y1)
plt.show()
```

## histogram

```python
x = np.random.normal(loc=80, scale=10, size=100)

plt.hist(x)
plt.show()
```

## subplots

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])
figure, axes = plt.subplots(2,2)

axes[0,0].plot(x, x*2, color='red')
axes[0,0].set_title('multiple x')

axes[0,1].plot(x, x**2, color='green')
axes[0,1].set_title('x**2')

axes[1,0].plot(x, x**3, color='pink')
axes[1,0].set_title('x**3')

axes[1,1].plot(x, x**4, color='blue')
axes[1,1].set_title('x**4')

plt.tight_layout()
plt.show()
```
